```python
import pandas as pd 
```

```python
insurance_data = pd.read_csv('medical_insurance.csv')
insurance_data.head()
```

```python
insurance_data.columns
```

```python
from sklearn.model_selection import train_test_split
insurance_data.dropna(subset=['annual_medical_cost'])
y = insurance_data['annual_medical_cost']
features = ["age","sex","region","urban_rural","income","education","employment_status","marital_status","household_size","dependents","bmi","smoker","alcohol_freq",
"hospitalizations_last_3yrs","plan_type","network_tier","deductible","copay","policy_term_years","hypertension","diabetes","asthma","copd","cardiovascular_disease",
"cancer_history","kidney_disease","liver_disease","arthritis","mental_health"]

x = insurance_data[features]
train_x,val_x,train_y,val_y = train_test_split(x,y,random_state = 1)

numerical_col = [col for col in train_x.columns if train_x[col].dtype in ['int64', 'float64']]
categorical_col = [col for col in train_x.columns if  train_x[col].dtype =='object' ]

columns = numerical_col + categorical_col
train_x = train_x[columns].copy()
val_x = val_x[columns].copy()
x = insurance_data[columns].copy()
```

```python
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error
from sklearn.ensemble import RandomForestRegressor
validated_model = RandomForestRegressor(random_state=1,max_leaf_nodes=100, min_samples_leaf=8,max_depth=10,n_estimators=100)
validated_model.fit(train_x,train_y)
value = validated_model.predict(val_x)
mae = mean_absolute_error(value,val_y)
mape = mean_absolute_percentage_error(value,val_y)

print("Valididated model performance after tuning :" ,mae , mape)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder


numerical_transformer = SimpleImputer(strategy = 'median')

catergorical_transformer = Pipeline(steps=[

    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('onehot',OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num',numerical_transformer,numerical_col),
    ('cat',catergorical_transformer,categorical_col)
])
pipeline = Pipeline(steps=
    [('preprocessor',preprocessor),
    ('model',validated_model)]
)
pipeline.fit(train_x,train_y)
mae = mean_absolute_error(pipeline.predict(val_x),val_y)
print(mae)

```

```python
import joblib
pipeline.fit(x,y)
joblib.dump(pipeline, 'medical_cost_pipeline_v1.joblib')

print("Pipeline saved successfully!")
```
